## Objective
Create bar and line charts to visualize monthly sales and identify trends.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('sales_data.csv')
display(df.head())
```

## 1. Inspect the dataset

```python
print('Dataset shape:', df.shape)
print('Columns:', df.columns.tolist())
print(df['Sales'].describe())
```

## 2. Bar chart — compare monthly sales

```python
plt.figure(figsize=(9, 5))
plt.bar(df['Month'], df['Sales'])
plt.title('Monthly Sales')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.tight_layout()
plt.show()
```

## 3. Line chart — observe the trend

```python
plt.figure(figsize=(9, 5))
plt.plot(df['Month'], df['Sales'], marker='o')
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## 4. Observations

```python
highest = df.loc[df['Sales'].idxmax()]
lowest = df.loc[df['Sales'].idxmin()]
print(f"Highest sales: {highest['Month']} ({highest['Sales']})")
print(f"Lowest sales: {lowest['Month']} ({lowest['Sales']})")
print('Total annual sales:', df['Sales'].sum())
print('Average monthly sales:', round(df['Sales'].mean(), 2))
```

## Conclusion

The bar chart helps compare individual months, while the line chart highlights the overall direction. Sales generally rise over the year, with small dips in March, June, and September. December is highest and January is lowest. These are illustrative sample values, not real business sales.
